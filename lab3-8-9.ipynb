```python
!pip install -U google-genai
from google import genai
client = genai.Client(api_key="")
MODEL = "models/gemini-3.5-flash"
def song_recommender(user_input):
    prompt = f"""
You are a music recommendation expert.
Provide a short, well-structured response with clear headings:
1. Mood/Vibe
2. Recommended Songs (3-5, Title - Artist)
3. Where to Listen (Spotify/YouTube search terms)
Keep it short and easy to read.
User request: {user_input}
Song Recommendations:
"""
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )
    return response.text
print("Gemini Song Recommender Started (type 'exit' to stop)\n")
while True:
    user = input("You: ")
    if user.lower() == "exit":
        print("Goodbye ")
        break
    try:
        reply = song_recommender(user)
        print("\nAgent:\n", reply, "\n")
    except Exception as e:
        print("Error:", e)
```
