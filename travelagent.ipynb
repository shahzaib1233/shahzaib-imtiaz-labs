```python
print("--- FLIGHT FINDER ---")

fares = {
    "Karachi": 150000,
    "Lahore": 110000,
    "Islamabad": 95000
}

city = input("Where would you like to fly?: ").lower().strip()
budget = int(input("What is your maximum spend?: "))

if city in fares:
    ticket_price = fares[city]

    if budget >= ticket_price:
        print("Agent: Success! You can afford a ticket to", city.capitalize())
    else:
        print("Agent: That flight exceeds your budget limit.")
        print("Agent: Try searching for a more budget-friendly route.")
else:
    print("Agent: Sorry, we do not fly to that destination.")
```
