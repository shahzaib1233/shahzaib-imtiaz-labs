```python
!pip install -U openai

from openai import OpenAI

client = OpenAI(
    api_key="sk-2720010c1660442481b601bb06e9369d",
    base_url="https://api.deepseek.com"
)

MODEL = "deepseek-chat"


# ---------------------------------------------------------
# Dictionary for info that CHANGES often (schedules, dates, etc.)
# Just add/update entries here whenever something changes.
# Format: "keyword to match in user question": "current info"
# ---------------------------------------------------------
SPECIAL_INFO = {
    "class schedule": "Update this with current semester's class schedule.",
    "admission deadline": "Update this with current admission deadline.",
    "exam dates": "Update this with current exam dates.",
}


def get_special_context(user_input):
    """Check if user's question matches any dynamic-info keyword, and pull it in."""
    matches = []
    for key, value in SPECIAL_INFO.items():
        if key.lower() in user_input.lower():
            matches.append(f"- {key}: {value}")
    return "\n".join(matches)


def faust_agent(user_input):
    special_context = get_special_context(user_input)

    system_prompt = f"""
You are the official information assistant for Federal Urdu University of Arts, Science and Technology (FAUST), Pakistan.

Answer student questions about departments, admissions, campus, fees, faculty, and general university info clearly and directly.

{"Latest known context (use this if relevant):\n" + special_context if special_context else ""}
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_input}
        ]
    )

    return response.choices[0].message.content


print("FAUST Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = faust_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
